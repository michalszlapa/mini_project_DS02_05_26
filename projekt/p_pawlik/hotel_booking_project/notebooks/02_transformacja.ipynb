```python
import pandas as pd
import numpy as np

df = pd.read_csv("../data/hotel_bookings_clean.csv")

df.head()
```

```python
df.shape
```

Na podstawie istniejących zmiennych utworzone zostaną nowe zmienne, które ułatwią analizę zależności pomiędzy cechami rezerwacji a jej anulowaniem.

```python
df["lead_time_group"] = pd.cut(
    df["lead_time"],
    bins=[-1, 7, 30, 90, 180, 365, float("inf")],
    labels=["0-7 dni", "8-30 dni", "31-90 dni", "91-180 dni", "181-365 dni", "365+ dni"]
)
```

```python
df["lead_time_group"].value_counts().sort_index()
```

```python
df[["lead_time", "lead_time_group"]].head(10)
```

```python
df["lead_time_group"].isna().sum()
```

### Grupowanie liczby specjalnych życzeń

Na podstawie zmiennej `total_of_special_requests` utworzona zostanie nowa zmienna `special_requests_group`, która grupuje rezerwacje według liczby specjalnych życzeń klienta.

```python
df["special_requests_group"] = pd.cut(
    df["total_of_special_requests"],
    bins=[-1, 0, 1, 2, 3, float("inf")],
    labels=["0", "1", "2", "3", "4+"]
)
```

```python
df["special_requests_group"].value_counts().sort_index()
```

```python
df.to_csv("../data/hotel_bookings_transformed.csv", index=False)
```
