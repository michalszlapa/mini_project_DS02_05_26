## Wnioski z analizy ilości braków w zbiorze danych

W zbiorze danych występują brakujące wartości w czterech kolumnach:
- `children` – 4 brakujące wartości,
- `country` – 488 brakujących wartości,
- `agent` – 16 340 brakujących wartości,
- `company` – 112 593 brakujące wartości.


W zbiorze danych nie wykryto żadnych duplikatów, dlatego nie ma konieczności ich usuwania.

```python
cancel_percent = (df["is_canceled"].value_counts(normalize=True) * 100).round(2)

print(f"Nieanulowane rezerwacje: {cancel_percent[0]}%")
print(f"Anulowane rezerwacje: {cancel_percent[1]}%")
```

```python
missing_percent = (df.isnull().sum() / len(df) * 100).round(2)

print("Kolumny zawierające brakujące wartości:\n")

for column, percent in missing_percent.items():
    if percent > 0:
        print(f"{column}: {percent}%")
```

```python
df["company"].value_counts(dropna=False)
```

```python
df.groupby(df["company"].notnull())["is_canceled"].mean()
```

Kolumna "company" zawiera 94,31% brakujących wartości. Dodatkowo przechowuje jedynie identyfikatory firm, bez informacji opisowych. Pomimo tego, że rezerwacje firmowe były rzadziej anulowane, bardzo wysoki odsetek braków oraz brak dodatkowych informacji o firmach sprawiają, że kolumna nie będzie wykorzystywana w dalszej analizie

```python
df.drop(columns=["company"], inplace=True)
```

```python
df.columns
```

```python
missing = df.isnull().sum()
missing[missing > 0]
```

```python
df[df["children"].isnull()]
```

```python
df[df["children"].isnull()][["children", "adults", "babies", "country", "agent"]]
```

```python
df.dropna(subset=["children"], inplace=True)
```

```python
missing = df.isnull().sum()
missing[missing > 0]
```

W kolumnie children znaleziono jedynie 4 brakujące wartości (0,003% wszystkich rekordów). Z uwagi na ich bardzo małą liczbę oraz brak możliwości wiarygodnego uzupełnienia, rekordy te zostały usunięte z dalszej analizy.

```python
df[df["country"].isnull()][
    ["hotel", "country", "market_segment", "distribution_channel", "agent"]
]
```

```python
df[df["country"].isnull()]["hotel"].value_counts()
```

```python
df.groupby(df["country"].isnull())["is_canceled"].mean()
```

```python
df.dropna(subset=["country"], inplace=True)
```

```python
missing = df.isnull().sum()
missing[missing > 0]
```

```python
df["agent"].value_counts()
```

```python
df.groupby(df["agent"].isnull())["is_canceled"].mean()
```

```python
df[df["agent"].isnull()][["market_segment", "distribution_channel"]].value_counts()
```

```python
df["agent"] = df["agent"].fillna(0)
```

```python
missing = df.isnull().sum()
missing[missing > 0]
```

Podsumowanie czyszczenia danych
Usunięto kolumnę company, ponieważ zawierała ponad 94% brakujących wartości.
Usunięto 4 rekordy z brakującą wartością w kolumnie children (0,003% danych).
Usunięto 488 rekordów z brakującą wartością w kolumnie country (0,41% danych), ponieważ nie było możliwości ich wiarygodnego uzupełnienia.
W kolumnie agent brakujące wartości zastąpiono wartością 0, ponieważ analiza wykazała, że NaN najprawdopodobniej oznacza brak pośrednika, a nie brak danych.
