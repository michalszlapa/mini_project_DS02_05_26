```python
data = data.dropna(subset='sales')   #skoro przewiduję SALES, to usuwam wiersze z n/a
```

```python
data.isnull().sum()
```

TRANSFORMACJA

```python
#wszystkie nuli w wydatkach oznaczają, że w tym tygodniu mój wydatek był 0


data_z_0 = data.copy()
data_z_0['instagram_spend'] = data_z_0['instagram_spend'].fillna(0)
data_z_0['google_ads_spend'] = data_z_0['google_ads_spend'].fillna(0)
data_z_0['tv_spend'] = data_z_0['tv_spend'].fillna(0)
data_z_0['youtube_spend'] = data_z_0['youtube_spend'].fillna(0)
data_z_0['newspaper_spend'] = data_z_0['newspaper_spend'].fillna(0)
data_z_0['influencer_spend'] = data_z_0['influencer_spend'].fillna(0)
data_z_0['ott_spend'] = data_z_0['ott_spend'].fillna(0)
```

```python
#zostały competitor spend. Uważam, że trzeba uzupełnić medianą, bo na pewno coś w tym momencie wydali, ale nie wiem ile

data_clear = data_z_0.copy()
data_clear['competitor_spend'] = data_clear['competitor_spend'].fillna(data_clear['competitor_spend'].median())
```

```python
print(data_clear.isna().sum())

data_clear.to_csv(r"data\data_clear")
```
