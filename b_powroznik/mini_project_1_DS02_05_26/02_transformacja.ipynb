```python
# Importowanie potrzebnych bibliotek
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('../mini_project_1_DS02_05_26/dataset/credit_risk_dataset_clean.csv')

df.head()
```

```python
# Korekta typów kolumn - zmiana teksowych na kategoryczne 

kolumny_kategorii = [
    "person_home_ownership",
    "loan_intent",
    "loan_grade",
    "cb_person_default_on_file",
]

for col in kolumny_kategorii:
  if col in df.columns:
    df[col] = df[col].astype("category")

df.dtypes
```

```python
# Mapowanie kolumny loan_grade (klasa ryzyka)  A - najniższe, G - najwyższe

grade_mapping = {"A": 0, "B": 1, "C": 2, "D": 3, "E": 4, "F": 5, "G": 6}
if "loan_grade" in df.columns:
  df["loan_grade_encoded"] = df["loan_grade"].map(grade_mapping)

# Opcjonalnie usuwamy oryginalną kolumnę tekstową
df = df.drop(columns=["loan_grade"])
```

```python
# Mapowanie kolumny cb_person_default_on_file  (histroia w BIK)  N - 0 , Y - 1


df["cb_person_default_on_file"] = df["cb_person_default_on_file"].map({"N": 0, "Y": 1})
```

```python
# Mapownie One-Hot Encoding dla kolumn loan_intent (Cel kredytu) oraz person_home_ownership (Status mieszkaniowy)

nominal_cols = ["loan_intent", "person_home_ownership"]
existing_nominal = [col for col in nominal_cols if col in df.columns]
if existing_nominal:
  df = pd.get_dummies(
      df, columns=existing_nominal, drop_first=True, dtype=int
  )

```

```python
# Tworzenie nowej cechy log_person_income (Logarytm dochodu)

df["log_person_income"] = np.log1p(df["person_income"])
```

```python
# Tworzenie nowej cechy cred_hist_to_age (Stosunek historii kredytowej do wieku)

df["cred_hist_to_age"] = df["cb_person_cred_hist_length"] / (
    df["person_age"] + 1
)

```

```python
#Zapisujemy plik po transpormacji

df.to_csv("../mini_project_1_DS02_05_26/dataset/credit_risk_dataset_transformed.csv", index=False)
```
