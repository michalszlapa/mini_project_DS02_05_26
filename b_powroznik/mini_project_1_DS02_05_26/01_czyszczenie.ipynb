```python
# Importowanie potrzebnych bibliotek
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('../mini_project_1_DS02_05_26/dataset/credit_risk_dataset.csv')
```

```python
df.head()
```

```python
# Sprawdzenie typu danych i ich struktury
df.info()
```

```python
#Statystyki kolumn numerycznych
df.describe()
```

```python
#Sprawdzenie brakujących wartości
missing = df.isna().sum().to_frame('braki')
missing['procent'] = (missing['braki'] / len(df) * 100).round(1)
missing.sort_values('braki', ascending=False)
```

Braki w zbiorze danych występują tylko w dwóch kolumanch:
`person_emp_length` oraz `loan_int_rate`


Wiersze z brakami `person_emp_length` zostaną usunięte


Wierszse z brakami `loan_int_rate` uzupełnimy medianą

```python
#Usunięcie wierszy z brakami w kolumnie długości zatrudnienia
df.dropna(subset=['person_emp_length'], inplace=True)
```

```python
#uzupełnienie medianą braków w kolumnie oprocentowania pożyczki
df['loan_int_rate'] = df['loan_int_rate'].fillna(df['loan_int_rate'].median())
df.dropna(inplace=True)
```

```python
#Sprawdzenie zduplikowanych wierszy
print(df.duplicated().sum())
```

W zbiorze znajduje się 165 zduplikowany wierszy, które usuniemy

```python
#Usunięcie zduplikowanych wierszy
df.drop_duplicates(inplace=True)
```
